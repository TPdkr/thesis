# YOLO26n

This notebook explores the YOLO library functions and custom functions created based on them. 

```python
import cv2
import time
import numpy as np
import os
import matplotlib.pyplot as plt 
from ultralytics import YOLO
from yolo import getEmbeds1, getEmbeds2, CONFIDENCE, CLASSES_YOLO, getEmbedFromResults
from kitty import listPicsWith, KITTY_PATH, CAR_YOLO_CLASSES, KITTY_CONF
```

```python
print("\nRESULTS FOR PICTURE: ")
#opening the image
test_path = "../datasets/depth_selection/val_selection_cropped/image/2011_09_26_drive_0005_sync_image_0000000146_image_02.png"

#custom function is called
start = time.time()
getEmbeds1([test_path], CONFIDENCE, CLASSES_YOLO, verbose=True)
end = time.time()
time1 = end-start

print("TEST1 FINSIHES---------")
start = time.time()
getEmbeds2([test_path], verbose=True)
end = time.time()
time2 = end-start

print("TIME: get crops took {0} to run".format(time1))
print("TIME: get crops 2 took {0} to run".format(time2))

#clean the directory
files = os.listdir("../imgs/crops/")
print("Deleting files: {0}".format(files))
for f in files:
    os.remove("../imgs/crops/{0}".format(f))
```

```python
images_with_cars = listPicsWith(KITTY_PATH, CAR_YOLO_CLASSES, KITTY_CONF, True)
```

```python
model = YOLO("../build/yolo26n.pt")

results = model.predict(images_with_cars, conf=KITTY_CONF, classes=CAR_YOLO_CLASSES)
print(f"Result type is :{type(results)} type of first item is {type(results[0])}")

start = time.time()
getEmbedFromResults(results, verbose=True)
end = time.time()
print(f"Time for getEmbedFromResults: {end-start}")
```
