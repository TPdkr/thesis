```python
import cv2
import time
import numpy as np
import os
import matplotlib.pyplot as plt 
from yolo import getEmbeds1, getEmbeds2, CONFIDENCE, CLASSES_YOLO
from kitty import listPicsWith, KITTY_PATH, CAR_YOLO_CLASSES, KITTY_CONF
```
